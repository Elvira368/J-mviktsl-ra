```python
import scipy.stats as stats
import matplotlib.pyplot as plt
import pandas as pd
hornsgatan = pd.read_csv("hornsgatan.csv")
period1 = hornsgatan.loc[(hornsgatan.period == "2008/09")]
print(period1)
m = period1.PM25.mean()
s = period1.PM25.std()
print("Medelvärde:", round(m, 4))
print("Standardavvikelse:", round(s, 4))
stats.probplot(period1.PM25, dist = "norm", sparams = [m, s], fit = True, plot = plt)
plt.title("Period 2008/09: Normalfördelning")
```

```python
param_logn = stats.lognorm.fit(period1.PM25)
print("Lognormal: ", param_logn)
param_logn0 = stats.lognorm.fit(period1.PM25, floc = 0)
print("Lognormal med loc=0: ", param_logn0)
fig, axs = plt.subplots(1, 2, constrained_layout = True)
stats.probplot(period1.PM25, dist = stats.lognorm, sparams = param_logn0,
fit = True, plot = axs[0])
stats.probplot(period1.PM25, dist = stats.lognorm, sparams = param_logn,
fit = True, plot = axs[1])
axs[0].set_title("Lognormal med loc = 0")
axs[1].set_title("Lognormal med loc = skattad")
```

```python
import numpy as np
import pandas as pd
hornsgatan = pd.read_csv("hornsgatan.csv")
# Period 1 = 2008/09
minperiod1 = hornsgatan.loc[(hornsgatan.period == "2008/09")]
# Antal observationer
n = minperiod1.shape[0]
# X
mx = minperiod1.PM25.mean()
sx = minperiod1.PM25.std()
# Y = ln(X)
my = np.log(minperiod1.PM25).mean()
sy = np.log(minperiod1.PM25).std()
# E(X) och D(X)
EX = np.exp(my + sy**2/2)
DX = EX * np.sqrt(np.exp(sy**2) - 1)
data = []
data.append([n, round(mx, 4), round(sx, 4), round(my, 4), round(sy, 4), round(EX, 4), round(DX, 4)])
df = pd.DataFrame(data, columns = ["n", "mx", "sx", "my", "sy", "E(X)", "D(X)"])
display(df)
```

```python
import numpy as np
import pandas as pd
hornsgatan = pd.read_csv("hornsgatan.csv")
period1 = hornsgatan.loc[(hornsgatan.period == "2008/09")]
n = period1.shape[0]
my = np.log(period1.PM25).mean()
sy = np.log(period1.PM25).std()
x05 = np.exp(my)
EX_logn = x05*np.exp(sy**2/2)
DX_logn = EX_logn*np.sqrt(np.exp(sy**2/2) - 1)
print("Lognormal: E(X) =", round(EX_logn, 4), ", D(X) =", round(DX_logn, 4))
d_norm = period1.PM25.std()/np.sqrt(n)
d_logn = DX_logn/np.sqrt(n)
print('d(barX): Normal =', d_norm, '; Lognormal =', d_logn)
```
